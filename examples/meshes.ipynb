# Example of loading and visualizing 3D meshes using ibl-datoviz

## Create an interactive datoviz app

```python
import datoviz as dvz
from ibl_datoviz.viewer import Viewer

app = dvz.App(background='white')
fig = app.figure(800, 600)
panel = fig.panel()
arcball = panel.arcball()
panel.update()
app.enable_ipython()
```

## Create an 3D brain viewer instance

```python
viewer = Viewer(app, panel)
```

## Add brain regions

```python
# Add single brain region
viewer.meshes.add_regions(['CA1'], hemisphere='left')
# Add list of brain regions
viewer.meshes.add_regions(['VISp', 'LGd', 'SCs'], hemisphere='left')
```

## Change transparency of brain regions


```python
# Change transparency of a single region (value between 0 and 255
viewer.meshes.set_alpha(100, 'CA1')
# Change transparency of all displayed regions
viewer.meshes.set_alpha(50)
```

## Hide regions


```python
viewer.meshes.hide_region('CA1')
```

## Show regions


```python
viewer.meshes.show_region('CA1')
```

## Set hemisphere


```python
viewer.meshes.set_hemisphere('both', 'CA1')
```

## Set colour



```python
viewer.meshes.set_color([30, 40, 50, 120], 'CA1')
```

## Reset colour


```python
viewer.meshes.reset_color('CA1')
```
